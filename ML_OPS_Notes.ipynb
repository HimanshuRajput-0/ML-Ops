It is highly challenging to automate and operationalize ML products and thus many ML endeavors fail to deliver on their expectations.

The paradigm of ML Ops addresses this issue.

MLOps includes several aspects, such as best practices, sets of concepts, and development culture.

In many industries data scientists still manage ML workflow manually to a great extent, resulting in many issues during the operations of the respective ML solution.

# Foundation of Devops:

Devops is more than a pure methodology and rather represents a paradigm addresing social and technical issues in organizations engaged in software development.

It has the goal of eliminating the gap between development and operations and emphasizes collaboration, communication and knowledge sharing.

It ensures automation with continuous integration, continuous delivery, and continuous deployment(CI/CD), thus allowing for fast, frequent, and reliable releases.

Moreover it is designed to ensure continuous testing, quality assurance, continuous monitoring, logging and feedback loops.

Devops tools are differentiated in 6 categories:

1. Collaboration and knowledge sharing.
2. Source Code management. (Github)
3. Build Process.
4. Continuous Integration.
5. Deployment automation. (Kubernetes, Dockers).
6. Monitoring and logging

Now because of Devops developers need to care about what they develop, as they need to operate it as well.

This same experience is now being used to automate and operationalize ML.

**Understanding: Devops automates every step or work and make it easy to manage and track the performance of the application or of a ML model.**
